## 2. Exploratory Data Analysis


```python
plt.figure(figsize=(10,5))
sns.heatmap(data.corr(),annot=True)
```

```python
#scatter plot
plt.scatter(data['YearsExperience'],data['Salary'])
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
```

## 3. Feature and Target Selection


```python
x=data['YearsExperience'] #features input data
y=data['Salary'] #target column
```

## 4. Train/Test Split


```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## 5. Linear Regression Model


```python
#model fitting
from sklearn.linear_model import LinearRegression
```

```python
LR=LinearRegression()
```

```python
#fit
LR.fit(x_train.values.reshape(-1,1),y_train)
```

```python
y_pred=LR.predict(x_test.values.reshape(-1,1))
```

## 6. Model Evaluation


```python
# prompt: mean absolute error

from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(y_test,y_pred))
```

```python
from sklearn import metrics
```

```python
r2=metrics.r2_score(y_test,y_pred)
```

```python
r2
```

## 7. Summary

This notebook applies Linear Regression to estimate salary from years of experience.  
The saved results report a **mean absolute error of about 2,446** and an **R² score of about 0.988**, indicating a very strong fit on this dataset.
